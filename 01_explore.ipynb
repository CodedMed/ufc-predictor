```python
import pandas as pd
df = pd.read_csv('data/ufc-master.csv')
df.shape
```

```python
df.columns.tolist()
```

```python
pd.set_option('display.max_columns', None)
df.columns.tolist()
```

```python
print(df['Winner'].value_counts())
print(df['date'].dtype)
```

```python
[c for c in df.columns if any(k in c.lower() for k in ['age', 'reach', 'height', 'td_', 'sig_str', 'win', 'stance'])]
```

```python
df['sig_str_pct_dif'] = df['R_avg_SIG_STR_pct'] - df['B_avg_SIG_STR_pct']
df['td_pct_dif'] = df['R_avg_TD_pct'] - df['B_avg_TD_pct']
[c for c in df.columns if 'losses' in c.lower()]
```

```python
df['losses_dif'] = df['R_losses'] - df['B_losses']
```

```python
df['odds_dif'] = df['R_odds'] - df['B_odds']

features = ['height_dif', 'reach_dif', 'age_dif', 'win_streak_dif',
            'longest_win_streak_dif', 'win_dif', 'sig_str_dif', 'avg_td_dif',
            'sig_str_pct_dif', 'td_pct_dif', 'losses_dif', 'odds_dif']

df[features].isnull().sum()
```

```python
df['sig_str_pct_dif'] = df['sig_str_pct_dif'].fillna(0)
df['td_pct_dif'] = df['td_pct_dif'].fillna(0)
df = df.dropna(subset=['odds_dif'])
```

```python
df.shape
```

```python
df = df[df['Winner'].isin(['Red', 'Blue'])]
df['target'] = (df['Winner'] == 'Red').astype(int)
df.shape
```

```python
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date').reset_index(drop=True)
df['date'].min(), df['date'].max()
```

```python
cutoff = '2024-01-01'
train = df[df['date'] < cutoff]
test = df[df['date'] >= cutoff]
print(train.shape, test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

X_train = train[features]
y_train = train['target']
X_test = test[features]
y_test = test['target']

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
model.score(X_test, y_test)
```

```python
from sklearn.metrics import log_loss, roc_auc_score

probs = model.predict_proba(X_test)[:, 1]
print("Log loss:", log_loss(y_test, probs))
print("ROC-AUC:", roc_auc_score(y_test, probs))
```

```python
import pandas as pd
pd.Series(model.coef_[0], index=features).sort_values()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model_scaled = LogisticRegression(max_iter=1000)
model_scaled.fit(X_train_scaled, y_train)

pd.Series(model_scaled.coef_[0], index=features).sort_values()
```

```python
features_no_odds = [f for f in features if f != 'odds_dif']
X_train2, X_test2 = train[features_no_odds], test[features_no_odds]

model2 = LogisticRegression(max_iter=1000)
model2.fit(X_train2, y_train)

print("Accuracy:", model2.score(X_test2, y_test))
probs2 = model2.predict_proba(X_test2)[:, 1]
print("Log loss:", log_loss(y_test, probs2))
print("ROC-AUC:", roc_auc_score(y_test, probs2))
```

```python
from sklearn.calibration import calibration_curve
import matplotlib.pyplot as plt

prob_true, prob_pred = calibration_curve(y_test, probs, n_bins=10)

plt.plot(prob_pred, prob_true, marker='o', label='Logistic Regression')
plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Perfect calibration')
plt.xlabel('Predicted probability')
plt.ylabel('Actual win rate')
plt.legend()
plt.title('Calibration: Logistic Regression')
plt.show()
```

```python
from xgboost import XGBClassifier

xgb = XGBClassifier(n_estimators=200, max_depth=3, learning_rate=0.1, random_state=42, eval_metric='logloss')
xgb.fit(X_train, y_train)

xgb_probs = xgb.predict_proba(X_test)[:, 1]

print("Accuracy:", xgb.score(X_test, y_test))
print("Log loss:", log_loss(y_test, xgb_probs))
print("ROC-AUC:", roc_auc_score(y_test, xgb_probs))
```

```python
def build_fighter_db(df):
    r = df[['R_fighter', 'date', 'R_Height_cms', 'R_Reach_cms', 'R_age',
            'R_current_win_streak', 'R_longest_win_streak', 'R_wins', 'R_losses',
            'R_avg_SIG_STR_landed', 'R_avg_SIG_STR_pct', 'R_avg_TD_landed', 'R_avg_TD_pct']].copy()
    r.columns = ['fighter', 'date', 'height', 'reach', 'age', 'win_streak',
                 'longest_streak', 'wins', 'losses', 'sig_str', 'sig_str_pct', 'td', 'td_pct']

    b = df[['B_fighter', 'date', 'B_Height_cms', 'B_Reach_cms', 'B_age',
            'B_current_win_streak', 'B_longest_win_streak', 'B_wins', 'B_losses',
            'B_avg_SIG_STR_landed', 'B_avg_SIG_STR_pct', 'B_avg_TD_landed', 'B_avg_TD_pct']].copy()
    b.columns = r.columns

    both = pd.concat([r, b])
    both = both.sort_values('date').drop_duplicates('fighter', keep='last')
    return both.set_index('fighter')

fighter_db = build_fighter_db(df)
fighter_db.to_pickle('fighter_db.pkl')
fighter_db.head()
```
